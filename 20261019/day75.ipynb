```python
import pandas as pd
df = pd.DataFrame({
    "姓名": ["小明","小红","小刚","小李","小王"],
    "班级": ["A","A","B","B","B"],
    "成绩": [92, 88, 76, 95, None],
    "年龄": [18, 17, 19, 18, 20],
}, index=[10, 20, 30, 40, 50])
print(df["姓名"])
print(df[["姓名", "成绩"]])
print(df.姓名)
```

```python
print(df.loc[10])
print(df.loc[10:30])
print(df.iloc[0])
print(df.iloc[0:2])
print(df.loc[10:30,["姓名", "成绩"]])
print(df.iloc[[0,2,4],1:3])
```

```python
m = (df["成绩"] >= 90) & (df["班级"] == 'A')
print(df[m])
print(df.query("成绩 >= 90 and 班级 == 'A'"))
top = df.nlargest(3,"成绩")
print(top)
bot = df.nsmallest(2,"年龄")
print(bot)
```

```python
df.loc[df["班级"] == 'B','成绩'] = df.loc[df['班级'] == 'B','成绩'].fillna(0)
print(df)
df["总分"] = df["成绩"] * 10
print(df)
df.insert(1,"性别", ["男","女","男","女","男"])
print(df)
```

```python
tmp = df[df["班级"] == 'A']
print(tmp)
tmp["成绩"] = 100
print(df)
df.loc[df["班级"] == 'A','成绩'] = 100
print(df)
tmp = df[df["班级"] == 'A'].copy()
tmp['成绩'] = 100
print(tmp)
```

```python
df2 = df.set_index('姓名')
print(df2.loc['小明'])
print(df2.reset_index())
print(df.reset_index(drop=True))
print(df.reindex([10,99,20]))
print(df.sort_values(['班级','成绩'],ascending=[True,False],na_position='last'))
print(df.sort_index())
```

```python
import pandas as pd
raw = pd.DataFrame({
    "订单号": ["00123", "00124", "00125"],
    "金额": ["1,234.5", "99.0", "-"],
    "日期": ["2024-01-05", "2024/02/10", "unknown"],
    "是否会员": ["Y", "N", "Y"],
    "城市": ["北京", "上海", "北京"],
})
print(raw.dtypes)
raw["订单号"] = raw['订单号'].astype(str).str.zfill(6)
print(raw)
raw['金额'] = pd.to_numeric(raw['金额'].str.replace(",",""),errors="coerce")
print(raw)
raw['日期'] = pd.to_datetime(raw['日期'],errors='coerce')
print(raw)
print(raw['日期'].isna().sum(),"行日期解析失败")
```

```python
print(raw['城市'].nunique(),"/",len(raw))
raw['城市'] = raw['城市'].astype('category')
print(raw['城市'].cat.categories)
print(raw['城市'].cat.codes)
s = pd.Series([1,2,None],dtype='Int64')
print(s + 1)
df = raw.convert_dtypes()
print(df)
```
