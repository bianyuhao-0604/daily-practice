```python
#给你一个单链表的头节点 head ，请你判断该链表是否为回文链表。如果是，返回 true ；否则，返回 false 。
#输入：head = [1,2,2,1]
#输出：true
#输入：head = [1,2]
#输出：false
from typing import Optional
class ListNode:
    def __init__(self,val=0,next=None):
        self.val = val
        self.next = next
class Solution:
    def isPalindrome(self,head:Optional[ListNode]) -> bool:
        if not head:
            return True
        left = head
        def is_pal(right:Optional[ListNode]) -> bool:
            nonlocal left
            if right.next and not is_pal(right.next):
                return False
            if left.val != right.val:
                return False
def build_linked_list(values):
    dummy = ListNode()
    cur = dummy
    for v in values:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next
def print_linked_list(head):
    values = []
    cur = head
    while cur:
        values.append(cur.val)
        cur = cur.next
    return values
if __name__ == "__main__":
    solution = Solution()
    head1 = build_linked_list([1,2,2,1])
    result1 = solution.isPalindrome(head1)
    print("链表1:",print_linked_list(head1),"是否回文:",result1)
```

```python
#给你一个链表，删除链表的倒数第 n 个结点，并且返回链表的头结点。
#输入：head = [1,2,3,4,5], n = 2
#输出：[1,2,3,5]
#输入：head = [1], n = 1
#输出：[]
#输入：head = [1,2], n = 1
#输出：[1]
from typing import Optional
class ListNode:
    def __init__(self,val=0,next=None):
        self.val = val
        self.next = next
class Solution:
    def removeNthFromEnd(self,head:Optional[ListNode],n:int) -> Optional[ListNode]:
        left = right = dummy = ListNode(next=head)
        for _ in range(n):
            right = right.next
        while right.next:
            left = left.next
            right = right.next
        left.next = left.next.next
        return dummy.next
def build_linked_list(values):
    dummy = ListNode()
    cur = dummy
    for v in values:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next
def to_list(head):
    values = []
    cur = head
    while cur:
        values.append(cur.val)
        cur = cur.next
    return values
if __name__ == "__main__":
    solution = Solution()
    head1 = build_linked_list([1,2,3,4,5])
    result1 = solution.removeNthFromEnd(head1,2)
    print("用例1 删除倒数第2个:", to_list(result1))
```
