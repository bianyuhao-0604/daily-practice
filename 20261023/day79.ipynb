```python
import pandas as pd
df = pd.read_csv('data/learn_pandas.csv')
df_demo = df[['Grade','Name','Height','Weight']].set_index(['Grade','Name'])
print(df_demo)
print(df_demo.sort_values('Height').head())
print(df_demo.sort_values('Height',ascending=False).head())
```

```python
print(df_demo.sort_values(['Weight','Height'],ascending=[True,False]).head())
df_demo = df[['Height','Weight']]
def my_mean(x):
    res = x.mean()
    return res
df_demo.apply(my_mean)
```

```python
print(df_demo.apply(lambda x:x.mean()))
print(df_demo.apply(lambda x:x.mean(),axis=1).head())
print(df_demo.apply(lambda x:(x-x.mean()).abs()).mean())
```

```python
s = pd.Series([1,2,3,4,5])
roller = s.rolling(window=3)
print(roller)
print(roller.mean())
print(roller.sum())
```

```python
s2 = pd.Series([1,2,6,16,30])
print(roller.cov(s2))
print(roller.corr(s2))
print(roller.apply(lambda x:x.mean()))
```

```python
import pandas as pd
df = pd.read_csv('data/pokemon.csv')
print(df.head(3))
cols=['Total','HP','Attack','Defense','Sp. Atk','Sp. Def','Speed']
s = df[cols].sum(axis=1)
(s == df['Total']).all()
(s != df['Total']).sum()
```

```python
df = df.drop_duplicates(subset='#',keep='first').reset_index(drop=True)
print(df['Type 1'].nunique())
vc = df['Type 1'].value_counts()
print(vc.head(3),"\n")
print(vc.index[:3])
print(df[['Type 1','Type 2']].drop_duplicates().shape[0],"\n")
print(df[['Type 1','Type 2']].drop_duplicates())
```
