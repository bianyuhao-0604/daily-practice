```python
df = pd.DataFrame({
    " 姓名 ": [" 小明 ", "小红", "小明", "小李", None],
    "邮箱": ["a@x.com", "B@X.COM", "a@x.com", "invalid", "c@x.com"],
    "金额": [100, -5, 200, 150, 300],
    "标签": [["a", "b"], ["b"], [], ["c"], ["a"]],
})
df.columns = df.columns.str.strip().str.lower().str.replace(" ","_")
df = df.rename(columns={"email":"邮箱"})
df = df.rename(columns=str.lower)
df["姓名"] = df["姓名"].str.strip().str.replace(r"\s+"," ",regex=True)
df["域名"] = df["邮箱"].str.split("@").str[1]
print(df["邮箱"].str.contains(r"@.*\.[a-z]+",case=False,na=False))
print(df["姓名"].str.len())
print(df["姓名"].str.extract(r"(.)\w*"))
print(df.duplicated(subset=["姓名","邮箱"]).sum())
print(df.drop_duplicates(subset=["邮箱"],keep="first"))
```

```python
df = pd.DataFrame({
    " 姓名 ": [" 小明 ", "小红", "小明", "小李", None],
    "邮箱": ["a@x.com", "B@X.COM", "a@x.com", "invalid", "c@x.com"],
    "金额": [100, -5, 200, 150, 300],
    "标签": [["a", "b"], ["b"], [], ["c"], ["a"]],
})
df["金额"] = df["金额"].where(df["金额"] >= 0,0)
df["金额"] = df["金额"].clip(lower=0)
df["档位"] = pd.cut(df["金额"],bins=[0,50,150,1e6],labels=["低","中","高"])
df["分位"] = pd.qcut(df["金额"],q=4,labels=["Q1","Q2","Q3","Q4"],duplicates="drop")
df["邮箱合法"] = ~df["邮箱"].str.contains("invalid|@",na=True)
sales = pd.DataFrame({
    "门店": ["A","A","B","B"], "品类": ["水果","蔬菜","水果","蔬菜"],
    "2024": [100, 200, 150, 250], "2025": [120, 210, 160, 240],
})
long = sales.melt(id_vars=["门店","品类"],var_name="年份",value_name="销售额")
print(long)
wide = long.pivot_table(index=["门店","品类"],columns="年份",values="销售额",aggfunc="sum")
print(wide.reset_index())
print(wide.stack())
print(df.explode("标签"))
```
