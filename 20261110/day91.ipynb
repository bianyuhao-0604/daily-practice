```python
from collections import deque
dq = deque([1,2,3],maxlen=5)
dq.append(4)
dq.appendleft(0)
print(dq)
print(dq.pop(),dq.popleft())
print(len(dq))
q = deque(["t1","t2"])
print(q.popleft())
print(q)
window = deque(maxlen=3)
for x in [1,2,3,4,5]:
    window.append(x)
print(window)
```

```python
from collections import Counter
c = Counter("mississippi")
print(c)
print(c["s"],c["z"])
print(c.most_common(2))
words = "py很好py万岁py".split()
wc = Counter(words)
print(wc)
a,b = Counter(a=3,b=1),Counter(a=1,b=2)
print(a + b)
print(a - b)
print(Counter("listen") == Counter("silent"))
```

```python
from collections import defaultdict
groups = defaultdict(list)
pairs = [("AI","Ada"),("CV","Bob"),("AI","Cara")]
for grp,name in pairs:
    groups[grp].append(name)
print(dict(groups))
counter = defaultdict(int)
for ch in "abracadabra":
    counter[ch] += 1
print(dict(counter))
print(counter["z"])
```
