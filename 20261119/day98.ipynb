```python
def outer_func(a,b):
    def addition(a,b):
        return a + b
    add = addition(a,b)
    return add + 5
print(outer_func(5,10))
```

```python
def addition(num):
    if num:
        return num  + addition(num - 1)
    else:
        return 0
res = addition(10)
print(res)
```

```python
def maxnum(numbers:list):
    max_number = numbers[0]
    for c in numbers:
        if c > max_number:
            max_number = c
    return max_number
print(maxnum([4, 6, 8, 24, 12, 2]))
```

```python
def print_info(**kwargs):
    for k,v in kwargs.items():
        print(k,v)
print_info(name="Alice", age=30, city="New York")
```

```python
global_var = 10
def modify_variable():
    global global_var
    global_var = 20
print("Initial:",global_var)
modify_variable()
print("Modified:",global_var)
```

```python
def recur_factorial(n):
    if n <= 1:
        return 1
    else:
        return n * recur_factorial(n - 1)
num = 5
print(f"The factorial of {num} is {recur_factorial(num)}")
```
