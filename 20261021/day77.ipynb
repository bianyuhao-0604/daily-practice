```python
import pandas as pd
df = pd.DataFrame({
    "大区": ["东","东","南","南","东","南"],
    "品类": ["数码","服饰","数码","服饰","数码","数码"],
    "月份": pd.to_datetime(["2024-01","2024-01","2024-01","2024-02","2024-02","2024-02"]),
    "销售额": [100,80,120,90,150,130],
    "订单数": [5,4,6,3,7,6],
})
pt = df.pivot_table(
    index=["大区"],columns="月份",values="销售额",
    aggfunc=["sum","mean"],fill_value=0,margins=True,margins_name="合计"
)
print(pt)
pt2 = df.pivot_table(index="大区",columns="月份",values=["销售额","订单数"],
aggfunc="sum",fill_value=0)
print(pt2.swaplevel(axis=1).sort_index(axis=1))
```

```python
L = []
def my_func(x):
    return 2 * x
for i in range(5):
    L.append(my_func(i))
print(L)
print([my_func(i) for i in range(5)])
print([m+'_'+n for m in ['a','b'] for n in ['c','d']])
```

```python
L1,L2,L3 = list('abc'),list('def'),list('hij')
print(list(zip(L1,L2,L3)))
print(tuple(zip(L1,L2,L3)))
for i,j,k in zip(L1,L2,L3):
    print(i,j,k)
L = list('abcd')
for index,value in enumerate(L):
    print(index,value)
for index,value in zip(range(len(L)),L):
    print(index,value)
print(dict(zip(L1,L2)))
zipped = list(zip(L1,L2,L3))
print(zipped)
print(list(zip(*zipped)))
```
