```python
import pandas as pd
df = pd.read_csv('data/learn_pandas.csv')
print(df['School'].unique())
print(df['School'].nunique())
print(df['School'].value_counts())
df_demo = df[['Gender','Transfer','Name']]
print(df_demo)
print(df_demo.drop_duplicates(['Gender','Transfer']))
df_demo.drop_duplicates(['Gender','Transfer'],keep='last')
```

```python
print(df_demo.drop_duplicates(['Name','Gender'],keep=False).head())
print(df['School'].drop_duplicates())
print(df_demo.duplicated(['Gender','Transfer']).head())
print(df['School'].duplicated().head())
```

```python
print(df['Gender'].replace({'Female':0,'Male':1}).head())
print(df['Gender'].replace(['Female','Male'],[0,1]).head())
```

```python
s = pd.Series([-1,1.2345,100,-50])
print(s.where(s<0))
print(s.where(s<0,100))
print(s.mask(s<0))
print(s.mask(s<0,-50))
s_condition=pd.Series([True,False,False,True],index=s.index)
print(s.mask(s_condition,-50))
```

```python
s = pd.Series([-1,1.2345,100,-50])
print(s.round(2))
print(s.abs())
print(s.clip(0,2))
```

```python
import pandas as pd
import numpy as np
s = pd.Series([-10,0,50,100,150])
lower,upper=0,100
result = pd.Series(
    np.select([s < lower,s > upper],[-1,999],default=s),
    index=s.index
)
print(result)
```
