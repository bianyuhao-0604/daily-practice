```python
np.random.seed(0)
L1,L2 = ['A','B','C'],['a','b','c']
mul_index1 = pd.MultiIndex.from_product([L1,L2],names=('Upper','Lower'))
L3,L4 = ['D','E','F'],['d','e','f']
mul_index2 = pd.MultiIndex.from_product([L3,L4],names=('Big','Small'))
df_ex = pd.DataFrame(np.random.randint(-9,10,(9,9)),index=mul_index1,columns=mul_index2)
df_ex
```

```python
np.random.seed(0)
L1,L2,L3 = ['A','B'],['a','b'],['alpha','beta']
mul_index1 = pd.MultiIndex.from_product([L1,L2,L3],names=('Upper','Lower','Extra'))
L4,L5,L6 = ['C','D'],['c','d'],['cat','dog']
mul_index2 = pd.MultiIndex.from_product([L4,L5,L6],names=('Big','Small','Other'))
df_ex = pd.DataFrame(np.random.randint(-9,10,(8,8)),index=mul_index1,columns=mul_index2)
df_ex
```

```python
print(df_ex.swaplevel(0,2,axis=1).head())
print(df_ex.reorder_levels([2,0,1],axis=0).head())
```

```python
print(df_ex.droplevel(1,axis=1))
print(df_ex.droplevel([0,1],axis=0))
```

```python
print(df_ex.rename_axis(index={'Upper':'Changedrow_row'},columns={'Other':'Changed_Col'}).head())
print(df_ex.rename(columns={'cat':'not_cat'},level=2).head())
df_ex.rename(index=lambda x:str.upper(x),level=2).head()
```

```python
import numpy as np
import pandas as pd
df = pd.read_csv('data/learn_pandas.csv')
print(df.groupby('Gender')['Height'].median())
print(df.groupby(['School','Gender'])['Height'].mean())
condition = df.Weight > df.Weight.mean()
print(df.groupby(condition)['Height'].mean())
```

```python
item = np.random.choice(list('abc'),df.shape[0])
print(df.groupby(item)['Height'].mean())
print(df.groupby([condition,item])['Height'].mean())
print(df[['School','Gender']].drop_duplicates())
df.groupby([df['School'],df['Gender']])['Height'].mean()
```
