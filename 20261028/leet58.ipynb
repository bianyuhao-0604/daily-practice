```python
#给你一个链表，删除链表的倒数第 n 个结点，并且返回链表的头结点。
#输入：head = [1,2,3,4,5], n = 2
#输出：[1,2,3,5]
#输入：head = [1], n = 1
#输出：[]
#输入：head = [1,2], n = 1
#输出：[1]
from __future__ import annotations
class ListNode:
    def __init__(self,val=0,next:ListNode | None = None):
        self.val = val
        self.next = next
    def __repr__(self):
        return f"{self.val}"
class Solution:
    def removeNthFromEnd(self,head:ListNode | None,n:int) -> ListNode | None:
        left = right = dummy = ListNode(next=head)
        for _ in range(n):
            right = right.next
        while right.next:
            left = left.next
            right = right.next
        left.next = left.next.next
        return dummy.next
def build_list(vals):
    dummy = ListNode()
    cur = dummy
    for v in vals:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next
def to_list(head):
    res = []
    while head:
        res.append(head.val)
        head = head.next
    return res
head = build_list([1, 2, 3, 4, 5])
n = 2
print(to_list(Solution().removeNthFromEnd(head, n)))  
```

```python
#给定一个 已排序 的链表的头节点 head。
#删除原始链表中所有 重复 数字的节点，只留下 不同 的数字。
#返回 已排序 的链表。
#输入：head = [1,2,3,3,4,4,5]
#输出：[1,2,5]
#输入：head = [1,1,1,2,3]
#输出：[2,3]
from __future__ import annotations
class ListNode:
    def __init__(self,val=0,next: ListNode | None = None):
        self.val = val
        self.next = next
    def __repr__(self):
        return f"{self.val}"
class Solution:
    def deleteDuplicates(self, head: ListNode | None) -> ListNode | None:
        cur = dummy = ListNode(next=head)
        while cur.next and cur.next.next:
            val = cur.next.val
            if cur.next.next.val == val:
                while cur.next and cur.next.val == val:
                    cur.next = cur.next.next
            else:
                cur = cur.next
        return dummy.next
def build_list(vals):
    dummy = ListNode()
    cur = dummy
    for v in vals:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next
def to_list(head):
    res = []
    while head:
        res.append(head.val)
        head = head.next
    return res
head = build_list([1, 2, 3, 3, 4, 4, 5])
print(to_list(Solution().deleteDuplicates(head))) 
```
