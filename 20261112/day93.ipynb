```python
class ArrayStack:
    def __init__(self):
        self._data = []
    def push(self,x):
        self._data.append(x)
    def pop(self):
        if self.is_empty():
            raise IndexError("pop from empty stack")
        return self._data.pop()
    def peek(self):
        return self._data[-1]
    def is_empty(self):
        return len(self._data) == 0
    def __len__(self):
        return len(self._data)
    def __repr__(self):
        return "栈底 → " + str(self._data) + " → 栈顶"
st = ArrayStack()
for x in (1,2,3):
    st.push(x)
    print(st)
print(st.pop())
print(st)
print(st.peek())
print(len(st),st.is_empty())
```

```python
class ArrayQueue:
    def __init__(self,capacity=4):
        self._buf = [None] * capacity
        self._head = 0
        self._size = 0
    def enqueue(self,x):
        if self._size == len(self._buf):
            self._resize(len(self._buf) * 2)
        tail = (self._head + self._size) % len(self._buf)
        self._buf[tail] = x
        self._size += 1
    def dequeue(self):
        if self._size == 0:
            raise IndexError("dequeue from empty queue")
        x = self._buf[self._head]
        self._buf[self._head] = None
        self._head = (self._head + 1) % len(self._buf)
        self._size -= 1
        return x
    def _resize(self,cap):
        old = [self._buf[(self._head + i) % len(self._buf)] for i in range(self._size)]
        self._buf = old + [None] * (cap - self._size)
        self._head = 0
    def __len__(self):
        return self._size
    def __repr__(self):
        items = [self._buf[(self._head + i) % len(self._buf)] for i in range(self._size)]
        return f"队头 → {items} ← 队尾 内部数组: {self._buf}"
q = ArrayQueue(4)
for ch in "abc":
    q.enqueue(ch)
print(q)    
print(q.dequeue())
q.enqueue("d")
q.enqueue("e")
print(q)
print(len(q))
```

```python
def func(num:int):
    if num <= 1:
        return num
    result = 1
    for i in range(1,num+1):
        result *= i
    return result
print(func(5))
```
