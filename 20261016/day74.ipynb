```python
import pandas as pd
raw = pd.DataFrame({
    "订单号": ["00123", "00124", "00125"],
    "金额": ["1,234.5", "99.0", "-"],
    "日期": ["2024-01-05", "2024/02/10", "unknown"],
    "是否会员": ["Y", "N", "Y"],
    "城市": ["北京", "上海", "北京"],
})
print(raw.dtypes)
raw['订单号'] = raw['订单号'].astype(str).str.zfill(6)
raw['金额'] = pd.to_numeric(raw['金额'].str.replace(",",""),errors="coerce")
raw['日期'] = pd.to_datetime(raw['日期'],errors='coerce')
print(raw['日期'].isna().sum(),'行日期解析失败')
```

```python
print(raw['城市'].nunique(),"/",len(raw))
raw['城市'] = raw['城市'].astype("category")
print(raw['城市'].cat.categories)
print(raw['城市'].cat.codes)
```
