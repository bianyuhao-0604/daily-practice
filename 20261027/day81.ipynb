```python
import pandas as pd
import numpy as np
df = pd.read_csv('data/learn_pandas.csv',usecols=['School','Grade','Name','Gender','Weight','Transfer'])
df_demo=df.set_index('Name')
print(df_demo.head())
print(df_demo.iloc[1,1])
print(df_demo.iloc[[0,1],[0,1]])
print(df_demo.iloc[1:4,2:4])
print(df_demo.iloc[lambda x:slice(1,4)])
print(df_demo.iloc[(df_demo.Weight>80).values].head())
```

```python
print(df_demo.head())
print(df_demo.School.iloc[1])
print(df_demo.School.iloc[1:5:2])
print(df.query('((School == "A")&'
         ' (Grade == "Senior")&'
         ' (Weight > 70))|'
         '((School == "B")&'
         ' (Grade != "Senior")&'
         ' (Weight > 80))'))
df.query('Weight > Weight.mean()').head()
```

```python
print(df.query('(Grade not in ["Freshman","Sophomore"]) and'
         '(Gender == "Male")').head())
print(df.query('Grade == ["Junior","Senior"]').head())
low,high = 70,80
print(df.query('Weight.between(@low,@high)').head())
```
