```python
def top_k(nums,k):
    heap = nums[:k]
    heapq.heapify(heap)
    for x in nums[k:]:
        if x > heap[0]:
            heapq.heapreplace(heap,x)
    return sorted(heap,reverse=True)
print(top_k([7,1,9,3,8,2],3))
pq = []
heapq.heappush(pq,(2,0,'整理日志'))
heapq.heappush(pq,(1,1,'线上告警'))
print(heapq.heappop(pq))
merged = list(heapq.merge([1,4,7],[2,5,8],[3,6]))
print(merged)
```

```python
class ArrayStack:
    def __init__(self):
        self._data = []
    def push(self,x):
        self._data.append(x)
    def pop(self):
        if self.is_empty():
            raise IndexError("pop from empty stack")
        return self._data.pop()
    def peek(self):
        return self._data[-1]
    def is_empty(self):
        return len(self._data) == 0
    def __len__(self):
        return len(self._data)
    def __repr__(self):
        return "栈底 → " + str(self._data) + " → 栈顶"
st = ArrayStack()
for x in (1,2,3):
    st.push(x)
    print(st)
print(st.pop())
print(st)
print(st.peek())
print(len(st),st.is_empty())
```

```python
PAIR = {")":"(","]":"[","}":"{"}
def is_balanced(expr):
    st = []
    for ch in expr:
        if ch in "([{":
            st.append(ch)
        elif ch in PAIR:
            if not st or st.pop() != PAIR[ch]:
                return False
    return not st
print(is_balanced("{[()]}"))
```

```python
class ArrayQueue:
    def __init__(self,capacity=4):
        self._buf = [None] * capacity
        self._head = 0
        self._size = 0
    def enqueue(self,x):
        if self._size == len(self._buf):
            self._resize(len(self._buf)*2)
        tail = (self._head + self._size) % len(self._buf)
        self._buf[tail] = x
        self._size += 1
    def dequeue(self):
        if self._size == 0:
            raise IndexError("dequeue from empty queue")
        x = self._buf[self._head]
        self._buf[self._head] = None
        self._head = (self._head + 1) % len(self._buf)
        self._size -=1
        return x
    def _resize(self,cap):
        old = [self._buf[(self._head + i) % len(self._buf)] for i in range(self._size)]
        self._buf = old + [None] * (cap - self._size)
        self._head = 0
    def __len__(self):
        return self._size
    def __repr__(self):
        items = [self._buf[(self._head)]]
```

```python
class ListNode:
    def __init__(self,val,next=None):
        self.val = val
        self.next = next
class SinglyLinkedList:
    def __init__(self):
        self.head = None
        self.size = 0
    def prepend(self,val):
        self.head = ListNode(val,self.head)
        self.size += 1
    def append(self,val):
        if self.head is None:
            self.head = ListNode(val)
        else:
            cur = self.head
            while cur.next:
                cur = cur.next
            cur.next = ListNode(val)
        self.size += 1
    def remove(self,val):
        prev,cur = None,self.head
        while cur:
            if cur.val == val:
                if prev is None:
                    self.head = cur.next
                else:
                    prev.next = cur.next
                self.size -= 1
                return True
            prev,cur = cur,cur.next
        return False
    def __iter__(self):
        cur = self.head
        while cur:
            yield cur.val
            cur = cur.next
    def __len__(self):
        return self.size
    def __repr__(self):
        return "->".join(map(str,self)) +"-> None"
ll = SinglyLinkedList()
for x in (0,1,2,3):
    ll.append(x)
print(ll)
ll.prepend(-1)
print(ll)
print(ll.remove(2))
print(ll)
print(len(ll))
```

```python
def reverse_list(head):
    prev,cur = None,head
    while cur:
        nxt = cur.next
        cur.next = prev
        prev,cur = cur,nxt
    return prev

def find_mid(head):
    slow = fast = head
    while fast and fast.next:
        slow,fast = slow.next,fast.next.next
    return slow

def has_cycle(head):
    slow = fast = head
    while fast and fast.next:
        slow,fast = slow.next,fast.next.next
        if slow is fast:
            return True
    return False

def build(vals):
    head = None
    for v in reversed(vals):
        head = ListNode(v,head)
    return head

def show(head):
    parts,cur = [],head
    while cur:
        parts.append(str(cur.val))
        cur = cur.next
    print("->".join(parts)+"-> None")
h = build([1,2,3,4,5])
rh = reverse_list(h)
show(rh)
print(find_mid(rh).val)
n1,n2 = ListNode(1),ListNode(2)
n1.next,n2.next = n2,n1
print(has_cycle(n1))
print(has_cycle(build([7,8])))
```
