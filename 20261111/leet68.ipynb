```python
#给你一个 非严格递增排列 的数组 nums ，请你 原地 删除重复出现的元素，使每个元素 只出现一次 ，返回删除后数组的新长度。元素的 相对顺序 应该保持 一致 。然后返回 nums 中唯一元素的个数。
#考虑 nums 的唯一元素的数量为 k。去重后，返回唯一元素的数量 k。
#nums 的前 k 个元素应包含 排序后 的唯一数字。下标 k - 1 之后的剩余元素可以忽略。
#输入：nums = [1,1,2]
#输出：2, nums = [1,2,_]
#输入：nums = [0,0,1,1,1,2,2,3,3,4]
#输出：5, nums = [0,1,2,3,4,_,_,_,_,_]
class Solution:
    def removeDuplication(self,nums:list[int]) -> int:
        k = 1
        for i in range(1,len(nums)):
            if nums[i] != nums[i-1]:
                nums[k] = nums[i]
                k += 1
        return k
s = Solution()
print(s.removeDuplication(nums = [0,0,1,1,1,2,2,3,3,4]))
```

```python
#给定一个按 非递减顺序 排列的整数数组 nums，请你 原地 删除部分重复元素，使得每个不同的元素 最多出现两次。元素的 相对顺序 应保持 不变。
#由于在某些语言中无法改变数组的长度，因此必须将结果放在数组 nums 的前半部分。更准确地说，如果删除重复元素后有 k 个元素，那么 nums 的前 k 个元素应该保存最终结果。前 k 个元素之后的内容无关紧要。
#将最终结果放入 nums 的前 k 个位置后，返回 k。
#不要 为另一个数组分配额外空间。你必须使用 O(1) 的额外内存，通过 原地 修改输入数组 来完成此操作。
#输入：nums = [1,1,1,2,2,3]
#输出：5, nums = [1,1,2,2,3,_]
#输入：nums = [0,0,1,1,1,1,2,3,3]
#输出：7, nums = [0,0,1,1,2,3,3,_,_]
class Solution:
    def removeDuplicates(self,nums:list[int]) -> int:
        stack_size = 2
        for i in range(2,len(nums)):
            if nums[i] != nums[i-2]:
                nums[stack_size] = nums[i]
                stack_size += 1
        return min(stack_size,len(nums))
s = Solution()
print(s.removeDuplicates(nums = [0,0,1,1,1,1,2,3,3]))
```
